# 1. Что такое компьютерное зрение

Компьютерное зрение — область, связанная с автоматическим получением информации из изображений и видео.

Человек смотрит на кадр и видит смысл:

- человек;
- автомобиль;
- дорожный знак;
- пять коробок;
- движение.

Программа изначально не получает эти понятия. Она получает числа.

```text
Камера или файл
        ↓
Кадр
        ↓
Массив чисел
        ↓
Алгоритм
        ↓
Объекты, признаки и события
```

Основные задачи:

- **классификация** — определить, что изображено;
- **детекция** — найти объекты и их координаты;
- **сегментация** — определить, какие пиксели принадлежат объекту;
- **трекинг** — сопоставить один и тот же объект на разных кадрах.

## Цветное изображение как набор значений

Кадр шириной 1280 и высотой 720 пикселей может иметь форму:

```python
(720, 1280, 3)
```

Здесь:

- `720` — строки, то есть высота;
- `1280` — столбцы, то есть ширина;
- `3` — цветовые каналы.

Один пиксель можно представить тремя числами:

```python
[38, 124, 217]
```

Это ещё не «небо», «кожа» или «автомобиль». Это числовое описание цвета.

Сначала посчитайте самостоятельно, затем запустите ячейку.

## ✍️ Задание 1. Фильтрация детекции

Реализуйте функцию:

```python
accept_detection(
    class_name,
    confidence,
    target_class="person",
    min_confidence=0.7
)
```

Она должна вернуть `True`, только если:

- класс совпадает с `target_class`;
- уверенность не ниже `min_confidence`.

Примеры:

```python
accept_detection("person", 0.85)  # True
accept_detection("car", 0.90)     # False
accept_detection("person", 0.60)  # False
```

````python
# ВАШ КОД
def accept_detection(
    class_name: str,
    confidence: float,
    target_class: str = "person",
    min_confidence: float = 0.7,
) -> bool:
    if (class_name == target_class and confidence >= min_confidence):# Замените следующую строку своей реализацией.
      return True
    return False
````

<details>
<summary><strong>💡 Подсказка</strong></summary>

Понадобится одно логическое выражение с оператором `and`.

Проверьте одновременно:

```python
class_name == target_class
confidence >= min_confidence
```
</details>

````python
# ✅ ПРОВЕРКА ЗАДАНИЯ 1
tests = [
    (("person", 0.85), True),
    (("car", 0.90), False),
    (("person", 0.60), False),
    (("car", 0.10, "car", 0.1), True),
]

passed = 0

for arguments, expected in tests:
    actual = accept_detection(*arguments)
    if actual == expected:
        passed += 1
    else:
        print(
            f"Не пройден тест: accept_detection{arguments} "
            f"вернула {actual}, ожидалось {expected}"
        )

if passed == len(tests):
    print("✅ Все тесты пройдены.")
else:
    print(f"Пока пройдено тестов: {passed} из {len(tests)}.")
````

Решение находится в следующей свёрнутой ячейке. Сначала добейтесь прохождения тестов самостоятельно.

````python
# @title
def accept_detection_solution(
    class_name: str,
    confidence: float,
    target_class: str = "person",
    min_confidence: float = 0.7,
) -> bool:
    return class_name == target_class and confidence >= min_confidence
````

````python
# ВАШ КОД
count = 0
def count_above_threshold(values: list[int], threshold: int) -> int:
    count = 0
    # Замените следующую реализацию.
    for value in values:
      if value > threshold:
        count += 1
    return count
````

## ✍️ Задание 3. Сводка по детекциям

Реализуйте функцию:

```python
summarize_detections(objects, min_confidence)
```

Она должна вернуть словарь, в котором:

- ключ — имя класса;
- значение — количество объектов этого класса;
- учитываются только объекты с уверенностью не ниже `min_confidence`.

Пример:

```python
[
    {"class_name": "person", "confidence": 0.9},
    {"class_name": "car", "confidence": 0.8},
    {"class_name": "person", "confidence": 0.4},
]
```

при пороге `0.7` должно превратиться в:

```python
{"person": 1, "car": 1}
```

````python
# ВАШ КОД
def summarize_detections(
    objects: list[dict],
    min_confidence: float,
) -> dict:
    summary = {}
    for detected_object in objects:
      class_name = detected_object["class_name"]
      if detected_object["confidence"] >= min_confidence :
        summary[class_name] = summary.get(class_name, 0) + 1
    return summary
````

## ✍️ Задание 5. Нормализация массива

Реализуйте функцию:

```python
normalize_to_unit(values)
```

Она должна преобразовать числовой массив в диапазон `[0.0, 1.0]`.

Формула:

```text
(value - minimum) / (maximum - minimum)
```

Для массива, где все значения одинаковые, верните массив нулей той же формы.

````python
# ВАШ КОД
def normalize_to_unit(values: np.ndarray) -> np.ndarray:
    maximum = values.max()
    minimum = values.min()
    if maximum == minimum:
        return np.zeros_like(values, dtype=float)
    result = (values - minimum) / (maximum - minimum)
    return result
````

````python
# ✅ ПРОВЕРКА ЗАДАНИЯ 5
test_values = np.array([10, 20, 30], dtype=float)
actual = normalize_to_unit(test_values)
expected = np.array([0.0, 0.5, 1.0])

constant_actual = normalize_to_unit(np.array([7, 7, 7]))

if np.allclose(actual, expected) and np.allclose(
    constant_actual,
    np.zeros(3),
):
    print("✅ Все тесты пройдены.")
else:
    print("Пока решение требует исправления.")
    print("Получено для [10, 20, 30]:", actual)
    print("Ожидалось:", expected)
    print("Получено для одинаковых значений:", constant_actual)
````

````python
def normalize_to_unit_solution(values: np.ndarray) -> np.ndarray:
    minimum = values.min()
    maximum = values.max()

    if minimum == maximum:
        return np.zeros_like(values, dtype=float)

    return (values - minimum) / (maximum - minimum)
````

Результат оказался не `270`, потому что `uint8` не может его хранить.

Безопасная последовательность:

````python
# ВАШ КОД
def build_brightness_report(
    image: np.ndarray,
    threshold: int,
) -> tuple[np.ndarray, dict]:
    # Замените следующую реализацию.
    binary = np.where(image > threshold, 255, 0).astype(np.uint8)
    mask = image > threshold
    report = {
        "shape": image.shape,
        "dtype": str(image.dtype),
        "threshold": threshold,
        "minimum":int(image.min()),
        "maximum":int(image.max()),
        "mean":float(image.mean()),
        "bright_pixels": int(mask.sum()),
        "total_pixels": image.size,
        "bright_ratio":float(mask.sum() / image.size)}
    return binary, report
````

<details>
<summary><strong>💡 План решения</strong></summary>

1. Сформируйте маску `image > threshold`.
2. Получите бинарное изображение.
3. Посчитайте яркие пиксели через `np.count_nonzero(mask)`.
4. Общее количество пикселей находится в `image.size`.
5. Долю получите делением ярких пикселей на общее количество.
6. Минимум, максимум и среднее доступны как методы массива.
</details>

````python
# ✅ ПРОВЕРКА ИТОГОВОГО ЗАДАНИЯ
small_image = np.array([
    [0, 100, 101],
    [200, 50, 255],
], dtype=np.uint8)

binary_actual, report_actual = build_brightness_report(
    small_image,
    threshold=100,
)

binary_expected = np.array([
    [0, 0, 255],
    [255, 0, 255],
], dtype=np.uint8)

required_keys = {
    "shape",
    "dtype",
    "threshold",
    "minimum",
    "maximum",
    "mean",
    "bright_pixels",
    "total_pixels",
    "bright_ratio",
}

conditions = [
    np.array_equal(binary_actual, binary_expected),
    binary_actual.dtype == np.uint8,
    set(report_actual.keys()) == required_keys,
    report_actual.get("shape") == (2, 3),
    report_actual.get("dtype") == "uint8",
    report_actual.get("threshold") == 100,
    report_actual.get("minimum") == 0,
    report_actual.get("maximum") == 255,
    np.isclose(report_actual.get("mean", -1), small_image.mean()),
    report_actual.get("bright_pixels") == 3,
    report_actual.get("total_pixels") == 6,
    np.isclose(report_actual.get("bright_ratio", -1), 0.5),
]

if all(conditions):
    print("✅ Итоговое задание выполнено правильно.")
else:
    print("Пока не все требования выполнены.")
    print("Бинарное изображение:")
    print(binary_actual)
    print("Отчёт:")
    print(report_actual)
````

````python
def build_brightness_report_solution(
    image: np.ndarray,
    threshold: int,
) -> tuple[np.ndarray, dict]:
    mask = image > threshold
    binary = np.where(mask, 255, 0).astype(np.uint8)

    bright_pixels = int(np.count_nonzero(mask))
    total_pixels = int(image.size)

    report = {
        "shape": image.shape,
        "dtype": str(image.dtype),
        "threshold": threshold,
        "minimum": int(image.min()),
        "maximum": int(image.max()),
        "mean": float(image.mean()),
        "bright_pixels": bright_pixels,
        "total_pixels": total_pixels,
        "bright_ratio": bright_pixels / total_pixels,
    }

    return binary, report
````
